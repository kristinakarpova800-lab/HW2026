```python
import random

# 1. random.random() — повертає випадкове число з плаваючою комою від 0.0 до 1.0 (не включаючи 1.0)
print("random.random():", random.random())

# 2. random.randint(значення1, значення2) — повертає випадкове ціле число в діапазоні від значення1 до значення2 (обидві межі включені)
print("random.randint(1, 10):", random.randint(1, 10))

# 3. random.choice([]) — обирає випадковий елемент із переданого списку/послідовності
my_list = ["яблуко", "банан", "полуниця"]
print("random.choice():", random.choice(my_list))
```

```python
#завдання 1
students = [
    "Олександр Мельник",
    "Софія Шевченко",
    "Максим Ковальчук",
    "Анастасія Бойко",
    "Денис Романюк",
    "Вікторія Ткаченко",
    "Андрій Бондаренко",
    "Катерина Кравченко",
    "Дмитро Павлюк",
    "Ірина Савчук",
]

for i in range(4):
    random_student = random.choice(students)
    print(f"Спроба {i+1}: обрано студента -> {random_student}" )
```

```python
#завдання 2
num_to_choose = 3
selected_group = random.sample(students, num_to_choose)

names_str = ", ".join(selected_group)

print(f"Група з {num_to_choose} випадкових студентів: {names_str}")
```

```python
#завдання 3
def coin_game(flips_count):
    flips = ["орел", "решка"]
    results = {"орел": 0, "решка": 0}

    for _ in range(flips_count):
        result = random.choice(flips)
        results[result] += 1

    print(f"Результати для {flips_count} підкидань:")
    print(f"  Орел: {results['орел']} разів")
    print(f"  Решка: {results['решка']} разів\n")


coin_game(100)
coin_game(1000)
```

```python
#завдання 4
options = ["камінь", "ножиці", "папір"]

computer_choice = random.choice(options)
player_choice = random.choice(options)

print(f"Вибір гравця: {player_choice}")
print(f"Вибір комп'ютера: {computer_choice}")

if player_choice == computer_choice:
    print("Нічия! 🤝")
elif (player_choice == "камінь" and computer_choice == "ножиці") or \
     (player_choice == "ножиці" and computer_choice == "папір") or \
     (player_choice == "папір" and computer_choice == "камінь"):
    print("Переміг гравець! 🎉")
else:
    print("Переміг комп'ютер! 💻")
```
