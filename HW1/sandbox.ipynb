```python
def get_application_text(receiver_name: str, holidays: str, duration: str, date: str, name: str) -> str:
  return f"""

                                         {receiver_name}

                             Заява

  Прошу надати мені {holidays}, тривалістю {duration} календарних днів.

  Дата:{date}                  {name}

  """

receiver_name = input("Вкажіть посаду та ПІБ отримувача (в давальному відмінку): ")
holidays = input("Вкажіть вид відпустки(наприклад: щорічну основну/у зв'язку з навчанням): ")
duration = input("Вкажіть тривалість відпустки(лише кількість днів): ")
date = input("Вкажіть дату: ")
name = input("Вкажіть ваше ПІБ: ")

print(get_application_text(receiver_name, holidays, duration, date, name))
```
