```python
with open("./text.txt", mode="r", encoding="utf-8") as file:
    text = file.read()
```

```python
import re

words = re.findall(r"[a-яА-ЯіІїЇєЄґҐ']+", text)
words = [word.lower() for word in words]

print(words)
```

```python
def get_word_frequency_percent(words_list):
    words_len = len(words_list)
    words_count = {}
    if words_len == 0:
        return words_count

    for word in words_list:
        if word in words_count:
            words_count[word] = words_count[word] + (1 / words_len) * 100
        else:
            words_count[word] = (1 / words_len) * 100
    return words_count

words
```

```python
def filter_short_words(words_list):
    filtered_words = []
    for word in words_list:
        if len(word) > 1:
            filtered_words.append(word)
    return filtered_words


def get_shortest_words(words_list):
    valid_words = filter_short_words(words_list)
    unique_words = list(set(valid_words))
    
    if not unique_words:
        return []

    min_len = len(unique_words[0])
    shortest = []

    for word in unique_words:
        if len(word) < min_len:
            min_len = len(word)
            shortest = [word]
        elif len(word) == min_len:
            shortest.append(word)
            
    return shortest

print(get_shortest_words(words))
```

```python
def get_longest_words(words_list):
    unique_words = list(set(words_list))
    if not unique_words:
        return []

    max_len = 0
    longest = []

    for word in unique_words:
        if len(word) > max_len:
            max_len = len(word)
            longest = [word]
        elif len(word) == max_len:
            longest.append(word)
    return longest

print(get_longest_words(words))
```

```python
def get_unique_words_count(words_list):
    return len(set(words_list))

print(get_unique_words_count(words))
```

```python
def get_total_words_count(words_list):
    return len(words_list)

print(get_total_words_count(words))
```

```python
def get_used_letters(text_str):
    letters = re.findall(r"[а-яiїєґ]", text_str.lower())
    return sorted(list(set(letters)))

print(get_used_letters(text))
```

```python
def get_most_frequent_letter(text_str):
    letters = re.findall(r"[а-яiїєґ]", text_str.lower())
    if not letters:
        return []

    letter_count = {}
    for letter in letters:
        if letter in letter_count:
            letter_count[letter] += 1
        else:
            letter_count[letter] = 1

    max_freq = max(letter_count.values())
    
    most_freq_letters = [
        let for let, count in letter_count.items() if count == max_freq
    ]
    return most_freq_letters
```

```python
print(f"Загальна кількість слів: {get_total_words_count(words)}")
print(f"Кількість унікальних слів: {get_unique_words_count(words)}")
print(f"Найкоротші слова: {get_shortest_words(words)}")
print(f"Найдовші слова: {get_longest_words(words)}")
print(f"Використані літери: {get_used_letters(text)}")
print(f"Найчастіша літера: {get_most_frequent_letter(text)}")

print("ТОП-5 найчастіших слів")
words_count = get_word_frequency_percent(words)

sorted_words = sorted(words_count.items(), key=lambda pair: pair[1], reverse=True)

for word, frequency in sorted_words[:5]:
    print(f"{word}: {frequency:.2f}%")
```
